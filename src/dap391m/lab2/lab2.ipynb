```python
import hvplot.polars as hv
import polars as pl
```

# Lab 2

```python
df = pl.read_csv("lab2.csv")
df = df.rename({"": "Subject"})
df
```

```python
df_sorted = df.sort(by="Very interested", descending=True)
df_sorted
```

```python
plotter = hv.hvPlotTabularPolars(df_sorted)

plot = plotter.bar(
    x="Subject",
    # y=["Very interested", "Somewhat interested", "Not interested"],
    xlabel="Subject",
    ylabel="Interested",
    legend=True,
    height=540,
    width=1200
)

plot
```
